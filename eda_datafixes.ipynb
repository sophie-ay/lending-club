```python
df_pl = df_pl.with_columns([
    #take number of years and convert to float
    pl.col('emp_length')
        .str.extract(r'(\d+)')
        .cast(pl.Float64)
        .alias('emp_length_v1')
    
    #remove months from term
    ,pl.col('term')
        .str.extract(r'(\d+)')
        .cast(pl.Float64)
        .alias('term_months')
    
])
```

```python
df_pl.head()
```

```python
#encode home ownership
ho_map = {'MORTGAGE':1, 'RENT':2,'OWN':3,'NONE':4, 'ANY':5, 'OTHER':6}

#encode Grades
grades_map = {'A':1, 'B':2,'C':3,'D':4,'E':5,'F':6,'G':7}


df_pl = df_pl.with_columns([
    pl.col('home_ownership')
        .replace(ho_map)
        .cast(pl.Int8)
        .alias('homeowner_encoded')

    ,pl.col('grade')
        .replace(grades_map)
        .cast(pl.Int8)
        .alias('grade_encode')
])
```

```python
#creating ratio features
df_pl = df_pl.with_columns([
    pl.when(pl.col('annual_inc') == 0)
    .then(None)
    .otherwise(pl.col('loan_amnt') / pl.col('annual_inc'))
    .alias('loan_to_inc')

    ,pl.when(pl.col('annual_inc') == 0)
    .then(None)
    .otherwise(pl.col('installment') / (pl.col('annual_inc') /12) )
    .alias('installments_to_inc')

    ,pl.when(pl.col('revol_util').is_null() | (pl.col('revol_util') == 0))
    .then(None)
    .otherwise(pl.col('revol_bal') / pl.col('revol_util'))
    .alias('credit_utilisation')
])
```

```python
#imputing nulls
df_pl = df_pl.with_columns([
    pl.col('emp_length_v1').fill_null(-99999)

    ,pl.col('revol_util').fill_null(-99999)

    ,pl.col('credit_utilisation').fill_null(-99999)

])
```

```python
#Encode final target variable
df_pl = df_pl.with_columns(
    pl.col('loan_status')
    .map_elements(lambda x: 1 if x == 'Charged Off' else 0, return_dtype=pl.Int8)
    .alias('default_flag')
)
```

```python
def winsorise(df,col,lower_pct,higher_pct):
    arr = df[col].to_numpy()
    low = np.percentile(arr, lower_pct)
    high = np.percentile(arr, higher_pct)
    arr_clipped = np.clip(arr, low, high)
    df = df.with_columns(
        pl.Series(f"{col}_v2", arr_clipped)
    )
    return df
```

```python
df_pl = winsorise(df_pl, 'annual_inc', 1, 95)
df_pl = winsorise(df_pl, 'dti', 1, 95)
df_pl = winsorise(df_pl, 'loan_to_inc', 1, 95)
df_pl = winsorise(df_pl, 'installments_to_inc', 1, 95)
df_pl = winsorise(df_pl, 'credit_utilisation', 1, 95)
```
